### Data Cleaning
#### dropping the invalid columns and missing data

```python
new_df = df.drop(columns=["Data Validity Comment","Comment","Uo Units","AlogP","Compound Key","Ligand Efficiency BEI","Ligand Efficiency LE",	"Ligand Efficiency LLE","Ligand Efficiency SEI", "BAO Format ID","BAO Label","Assay Tissue ChEMBL ID","Assay Tissue Name","Assay Cell Type","Assay Subcellular Fraction","Assay Parameters","Assay Variant Accession","Assay Variant Mutation","Cell ChEMBL ID","Properties","Action Type","Standard Text Value","Value"])
```

```python
new_df.head(2)
```

```python
#to drop missing data
new2_df = new_df.dropna(subset=["Molecule ChEMBL ID","Molecule Name","Standard Type","Target ChEMBL ID","Target Organism"])
new2_df.head(2)
```

```python
#number of unique compounds by id
unique_cmpds = new2_df["Molecule ChEMBL ID"].nunique()
unique_cmpds
```

```python
#max cmpd potency
max_pchem_value = new2_df["pChEMBL Value"].max(axis=0)
max_pchem_value
```

```python
#min 
min_pchem_value = new2_df["pChEMBL Value"].min(axis=0)
min_pchem_value
```

```python
#mean
mean_pchembl = new2_df["pChEMBL Value"].mean()
mean_pchembl
```

```python
#ranking compounds by chembl
rank = new2_df.sort_values(by='pChEMBL Value',ascending= False)
rank.head(4)
```

```python
#sort by potency
top_10 = new2_df.sort_values(by='pChEMBL Value',ascending= False)
top_10.head()
```

```python
print(top_10[["Molecule Name","pChEMBL Value"]])
```

```python
#since Troclosan and chloroquine had duplicates
#group by the molecule and obtain max potency
top10_uniq = new2_df.groupby("Molecule Name")["pChEMBL Value"].max().reset_index()
top10_unique = top10_uniq.sort_values(by="pChEMBL Value",ascending=False)
top10_unique
```

## Visualization analysis 
#### a)by potency

```python
#bar chart
sns.barplot(x="Molecule Name",y="pChEMBL Value",data=top10_unique,palette='viridis',hue="Molecule Name",legend=False)
#rotation of the x-axis labels
plt.xticks(rotation=45)
plt.title("Top 10 Approved compounds by their reported potency(pChEMBL value)",pad=20,fontsize=15)
plt.xlabel("Compound")
plt.ylabel("Max pChEMBL Value")
plt.show()
```

```python
# Overwrite the existing top10_unique variable
idx = new2_df.groupby('Molecule Name')['pChEMBL Value'].idxmax()
top10_unique2 = new2_df.loc[idx].sort_values(by="pChEMBL Value", ascending=False).head(10)
top10_unique2.head(2)
```

```python
# Scatter plot
sns.scatterplot(x="Molecular Weight", y="pChEMBL Value", data=top10_unique2)

# Add titles and labels (matching your previous style)
plt.title("Molecular Weight vs. Reported Potency (pChEMBL value)", fontsize=15, pad=20)
plt.xlabel("Molecular Weight")
plt.ylabel("Max pChEMBL Value")

plt.show()
```
